```python
import pandas as pd
import html
import time
from pathlib import Path
```

```python
# PATH SETUP
RAW_DATA_FILE = "../data/raw/dataset.csv"
PROCESSED_DIR =  "../data/processed"
PROCESSED_DATA_FILE = PROCESSED_DIR + "/cleaned_sentiment140.csv"

print("Raw path:", RAW_DATA_FILE)
print("Processed path:", PROCESSED_DATA_FILE)
```

## Cleaning data

```python
# Load Data
start_time = time.time()

column_names = ['target', 'id', 'date', 'flag', 'user', 'text']
df = pd.read_csv(RAW_DATA_FILE, encoding='latin-1', names=column_names)

print(f"Dataset loaded in {time.time() - start_time:.2f} seconds.")
print(f"Total Rows: {df.shape[0]:,}")
print(f"Total Columns: {df.shape[1]}")
```

```python
# Keep Relevant Columns
df = df[['target', 'text']]
df.head()
```

```python
# Map Target Labels
print("Mapping target labels (4 -> 1)...")
df['target'] = df['target'].replace(4, 1)

df['target'].value_counts()
```

## Text Cleaning Pipeline

```python
print("Applying vectorized text cleaning...")
clean_start = time.time()

# A. Decode HTML
df['text'] = df['text'].apply(html.unescape)

# B. Lowercase
df['text'] = df['text'].str.lower()

# C. Replace URLs
df['text'] = df['text'].str.replace(r'http\S+|www\.\S+', '[URL]', regex=True)

# D. Replace mentions
df['text'] = df['text'].str.replace(r'@\w+', '[USER]', regex=True)

# E. Remove extra whitespace
df['text'] = df['text'].str.replace(r'\s+', ' ', regex=True).str.strip()

print(f"Cleaning time: {time.time() - clean_start:.2f} seconds")
```

```python
# Save Processed Data
df.to_csv(PROCESSED_DATA_FILE, index=False, encoding='utf-8')

print(f"Saved to: {PROCESSED_DATA_FILE}")
print(f"Total time: {time.time() - start_time:.2f} seconds")
```
